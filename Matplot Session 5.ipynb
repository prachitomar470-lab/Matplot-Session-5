```python

import matplotlib.pyplot as plt

scores = [150, 180, 165, 190, 175, 200, 145, 160, 185, 170,
          155, 195, 210, 180, 175, 165, 220, 140, 155, 190,
          200, 185, 170, 160, 150, 205, 195, 180, 175, 165,
          145, 155, 210, 200, 185, 170, 160, 190, 175, 180,
          150, 195, 205, 165, 155, 185, 200, 170, 180, 190]

plt.subplot(1, 3, 1)
plt.hist(scores, bins=5)
plt.title("5 Bins")

plt.subplot(1, 3, 2)
plt.hist(scores, bins=10)
plt.title("10 Bins")

plt.subplot(1, 3, 3)
plt.hist(scores, bins=15)
plt.title("15 Bins")

plt.show()
```

```python

import matplotlib.pyplot as plt

steps = [4000, 5000, 6000, 7000, 8000, 6500, 7500,
         4500, 5500, 6200, 7200, 8100, 6800, 7800,
         4200, 5200, 6100, 7100, 8300, 6600, 7600,
         4800, 5800, 6400, 7400, 8500, 6900, 7900,
         4300, 5300]

plt.subplot(1, 2, 1)
plt.hist(steps)
plt.title("Frequency")

plt.subplot(1, 2, 2)
plt.hist(steps, density=True)
plt.title("Density")

plt.show()
```

```python

import matplotlib.pyplot as plt
import random

songs = []

for i in range(50):
    songs.append(random.randint(120, 300))

plt.hist(songs, color="purple")
plt.title("Spotify Song Duration Distribution")
plt.show()
```

```python

import matplotlib.pyplot as plt

prices = [150, 250, 450, 600, 800, 950, 1200, 1500,
          1800, 2200, 3000, 4500, 350, 700, 2500]

bins = [100, 500, 1000, 2000, 5000]

plt.hist(prices, bins=bins)
plt.title("Flipkart Product Prices")
plt.xlabel("Price (Rs.)")
plt.ylabel("Number of Products")
plt.show()
```
